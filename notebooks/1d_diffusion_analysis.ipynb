## 2. Simulation Overview

This section provides a summary of the simulation data generated for the 1D PDE sandbox experiments.

We simulate two primary cases:
- **Reaction-Diffusion** with a linear reaction term
- **Advection-Diffusion** with a spatially varying drift

Each simulation tracks the distribution of particles over time under weighted Monte Carlo sampling.
Simulations are saved to structured directories, with parameter values embedded in folder names 
(e.g., 'T1.00_D1.0_alpha1.0_x0_0.0), and include '.npy' files for:
- 'positions': final particle locations
- 'weights': accumulated path weights

These results are compared to **analytical solutions** for each case to validate the accuracy of the Monte
Carlo solver.

```python
# Define the simulation data directories
reaction_dir = Path("../data/reaction")
advection_dir = Path("../data/advection")

# Parse folder names into DataFrames
reaction_df = parse_simulation_folder(reaction_dir)
advection_df = parse_simulation_folder(advection_dir)

# Helper function for clean printing
def print_parameter_summary(df, label=""):
    """
    Prints a clean summary of unique parameter values from a simulation DataFrame.
    """
    print(f"{label} Parameter Ranges:")
    for col in ["T", "D", "alpha", "x0"]:
        unique_vals = sorted(df[col].unique())
        formatted = ', '.join(f"{v:.2f}" if isinstance(v, float) else str(v) for v in unique_vals)
        print(f"  {col}: [{formatted}]")
    print()

# Print summary
print_parameter_summary(reaction_df, "Reaction")
print(f"Total Reaction Simulations: {len(reaction_df)}")
print_parameter_summary(advection_df, "Advection")
print(f"Total Advection Simulations: {len(advection_df)}")
```
